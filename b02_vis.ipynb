```python
import pandas as pd
import requests
import matplotlib.pyplot as plt
```

```python
# Fetch the data - gdp.
df_g = pd.read_csv("https://ourworldindata.org/grapher/real-gdp-growth.csv?v=1&csvType=full&useColumnShortNames=true", storage_options = {'User-Agent': 'Our World In Data data fetch/1.0'})
# Fetch the metadata
metadata_g = requests.get("https://ourworldindata.org/grapher/real-gdp-growth.metadata.json?v=1&csvType=full&useColumnShortNames=true").json()

# Fetch the data - domestic hours.
df_d = pd.read_csv("https://ourworldindata.org/grapher/share-of-day-women-spend-on-unpaid-domestic-and-care-work.csv?v=1&csvType=full&useColumnShortNames=true", storage_options = {'User-Agent': 'Our World In Data data fetch/1.0'})
# Fetch the metadata
metadata_d = requests.get("https://ourworldindata.org/grapher/share-of-day-women-spend-on-unpaid-domestic-and-care-work.metadata.json?v=1&csvType=full&useColumnShortNames=true").json()
```

```python
# Domestic Data
# Identify entity/country column and value column by position
entity_col = df_d.columns[0]  # First column (Entity)
year_col = df_d.columns[2]    # Third column (Year)
value_col = df_d.columns[3]   # Fourth column (women domestic labor)

# Select desired columns for chosen countries
selected_columns_d = [year_col, value_col]
brazil_df = df_d.loc[df_d[entity_col] == 'Brazil', selected_columns_d]
us_df = df_d.loc[df_d[entity_col] == 'United States', selected_columns_d]
switz_df = df_d.loc[df_d[entity_col] == 'Switzerland', selected_columns_d]
sk_df = df_d.loc[df_d[entity_col] == 'South Korea', selected_columns_d]

# GDP Data
# Identify entity/country column and value column by position
g_entity_col = df_g.columns[0]  # First column (Entity)
g_year_col = df_g.columns[2]    # Third column (Year)
g_value_col = df_g.columns[3]   # Fourth column (GDP growth)

# Select desired columns for chosen countries
selected_columns_dg = [g_year_col, g_value_col]
brazil_dg = df_g.loc[df_g[g_entity_col] == 'Brazil', selected_columns_dg]
us_dg = df_g.loc[df_g[g_entity_col] == 'United States', selected_columns_dg]
switz_dg = df_g.loc[df_g[g_entity_col] == 'Switzerland', selected_columns_dg]
sk_dg = df_g.loc[df_g[g_entity_col] == 'South Korea', selected_columns_dg]
```

```python
# Plot data

# Filter year ranges directly during plot calls
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Plot Domestic Unpaid Labor
ax1.plot(brazil_df[brazil_df[year_col].between(2003, 2024)][year_col], brazil_df[brazil_df[year_col].between(2003, 2024)][value_col], label='Brazil', marker='o')
ax1.plot(us_df[us_df[year_col].between(2003, 2024)][year_col], us_df[us_df[year_col].between(2003, 2024)][value_col], label='US', marker='o')
ax1.plot(switz_df[switz_df[year_col].between(2003, 2024)][year_col], switz_df[switz_df[year_col].between(2003, 2024)][value_col], label='Switzerland', marker='o')
ax1.plot(sk_df[sk_df[year_col].between(2003, 2024)][year_col], sk_df[sk_df[year_col].between(2003, 2024)][value_col], label='South Korea', marker='o')
ax1.set_title("Unpaid Domestic Labor (Hours/Day)")
ax1.legend()
ax1.grid(True)

# Plot GDP Growth
ax2.plot(brazil_dg[brazil_dg[g_year_col].between(2003, 2024)][g_year_col], brazil_dg[brazil_dg[g_year_col].between(2003, 2024)][g_value_col], label='Brazil', marker='s')
ax2.plot(us_dg[us_dg[g_year_col].between(2003, 2024)][g_year_col], us_dg[us_dg[g_year_col].between(2003, 2024)][g_value_col], label='US', marker='s')
ax2.plot(switz_dg[switz_dg[g_year_col].between(2003, 2024)][g_year_col], switz_dg[switz_dg[g_year_col].between(2003, 2024)][g_value_col], label='Switzerland', marker='s')
ax2.plot(sk_dg[sk_dg[g_year_col].between(2003, 2024)][g_year_col], sk_dg[sk_dg[g_year_col].between(2003, 2024)][g_value_col], label='South Korea', marker='s')
ax2.set_title("Real GDP Growth (%)")
ax2.legend()
ax2.grid(True)

plt.tight_layout()
plt.show()
```
