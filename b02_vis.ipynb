Data Visualization

```python
import pandas as pd
import requests
import matplotlib.pyplot as plt
```

```python
# Fetch the data.
df = pd.read_csv("https://ourworldindata.org/explorers/climate-change.csv?v=1&csvType=full&useColumnShortNames=true&Metric=Temperature+anomaly&Long-run+series=false", storage_options = {'User-Agent': 'Our World In Data data fetch/1.0'})

# Fetch the metadata
metadata = requests.get("https://ourworldindata.org/explorers/climate-change.metadata.json?v=1&csvType=full&useColumnShortNames=true&Metric=Temperature+anomaly&Long-run+series=false").json()
# Temperature Data
# Identify entity/country column and value column by position
t_col = df.columns[3]  # Fourth column (Entity)
year_col = df.columns[2]    # Third column (Year)
```

```python
# Add 13.7 to the temperature values
df[t_col] = df[t_col] + 13.7
# Filter for World
world_df = df[df[df.columns[0]] == 'World']

# plot
plt.figure(figsize=(10, 5))
plt.plot(world_df[year_col], world_df[t_col], label='World Absolute Temp (°C)')
# sneaky giant y-axis:
plt.ylim(0, 30)
plt.xlabel('Year')
plt.ylabel('Temperature (°C)')
plt.title('Absolute Temperature (Base + 13.7°C Anomaly)')
plt.legend()
plt.show()
```
